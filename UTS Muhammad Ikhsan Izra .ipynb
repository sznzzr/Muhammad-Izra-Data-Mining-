```python
combine = pd.concat([ddos_icmp,mqtt_dos,recon_port],ignore_index=True)
```

```python
combine.columns.values
```

```python
cmb = combine.iloc[:,7: 76]
```

```python
cmb
```

```python
cm = combine.iloc[:,83: 84]
```

```python
cm
```

```python
from sklearn.model_selection import train_test_split
```

```python
cmb_train, cmb_test, cm_train, cm_test = train_test_split(cmb, cm, test_size= 0.2, random_state = 42)
```

```python
from sklearn import tree
from sklearn.tree import DecisionTreeClassifier
```

```python
ucup = DecisionTreeClassifier(criterion='entropy', splitter = 'random')
ucup.fit(cmb_train,cm_train)
```

```python
cm_pred = ucup.predict(cmb_test)
```

```python
cm_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(cm_test,cm_pred)
```

```python
accuracy
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
fig = plt.figure(figsize = (10, 7))
tree.plot_tree(ucup, feature_names = cmb.columns.values, class_names = np.array([ 'DoS ICMP Flood','MQTT DoS Connect Flood','Recon Port Scan']), filled = True)
plt.show()
```

```python
import seaborn as mnm
from sklearn import metrics
label = np.array([ 'ddos_icmp','mqtt_dos','recon_port'])
```

```python
import matplotlib.pyplot as plt
conf_matrix = metrics.confusion_matrix(cm_test, cm_pred)
plt.figure(figsize=(14, 14))
mnm.heatmap(conf_matrix, annot=True, xticklabels=label, yticklabels=label)
plt.xlabel('Prediksi')
plt.ylabel('Fakta')
plt.show()
```
